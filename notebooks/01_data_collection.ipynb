# 01 · Data Collection

This notebook documents the data sourcing for MailMind AI's spam/ham classification pipeline.

## Dataset: UCI SMS Spam Collection v1

| Property | Value |
|----------|-------|
| **Source** | [UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/228/sms+spam+collection) |
| **License** | CC BY 4.0 |
| **Format** | Tab-separated (label \t text) |
| **Records** | 5,574 raw messages |
| **Classes** | ham (legitimate), spam |
| **Language** | English |
| **Citation** | Almeida, T.A., Gómez Hidalgo, J.M., Yamakami, A. (2011). Contributions to the Study of SMS Spam Filtering. |

### Why this dataset?
- Well-established benchmark in NLP/spam research
- Real-world messages (not synthetic)
- Imbalanced classes (~87% ham, ~13% spam) reflecting production reality
- Open license suitable for academic and commercial use

```python
import pandas as pd
from pathlib import Path

RAW_DIR = Path('../data/raw')
filepath = RAW_DIR / 'SMSSpamCollection'

df = pd.read_csv(filepath, sep='\t', header=None, names=['label', 'text'], encoding='latin-1')
print(f'Loaded {len(df):,} raw messages')
print(f'\nLabel distribution:')
print(df['label'].value_counts())
print(f'\nSample messages:')
df.head(10)
```

## Raw Data Inspection

```python
print('Data types:', df.dtypes.to_dict())
print(f'\nNull values: {df.isnull().sum().to_dict()}')
print(f'Duplicate texts: {df.duplicated(subset=["text"]).sum()}')
print(f'\nClass imbalance ratio: {df["label"].value_counts()["ham"] / df["label"].value_counts()["spam"]:.1f}:1 (ham:spam)')
```

## Data Integrity Checks

Before proceeding to cleaning, we verify:
1. No unexpected labels exist
2. No empty messages
3. Document the duplicate count for the cleaning step

```python
# Verify only expected labels
assert set(df['label'].unique()) == {'ham', 'spam'}, 'Unexpected labels found!'

# Check for empty messages
empty = df[df['text'].str.strip() == '']
print(f'Empty messages: {len(empty)}')

# Length statistics
df['length'] = df['text'].str.len()
print(f'\nMessage length stats:')
print(df.groupby('label')['length'].describe().round(1))

print('\n✅ Data collection verified. Proceed to 02_data_cleaning.')
```
